## 3.2 NumPy 数组

NumPy 的基本数据结构是可索引的 n 维*数组（array）*，其中的元素类型相同（`dtype`）。你可能马上就会注意到，我们把“维度”这个词用在了两个意思上。上面，它指向量中元素的个数；而这里，维度指数组的索引个数。一维（1-D）数组有一个索引。在课程 1 中，我们将用 NumPy 一维数组表示向量。

 - 1-D 数组，形状 (n,)：n 个元素，索引从 [0] 到 [n-1]

## 3.3 创建向量


NumPy 中的数据创建例程通常第一个参数是对象的形状。它可以是表示一维结果的单个值，也可以是形如 (n,m,...) 的元组，用于指定结果的形状。下面是使用这些例程创建向量的例子。

```python
# 分配内存并用指定值填充数组的 NumPy 程序
a = np.zeros(4);                print(f"np.zeros(4) :   a = {a}, a shape = {a.shape}, a data type = {a.dtype}")
a = np.zeros((4,));             print(f"np.zeros(4,) :  a = {a}, a shape = {a.shape}, a data type = {a.dtype}")
a = np.random.random_sample(4); print(f"np.random.random_sample(4): a = {a}, a shape = {a.shape}, a data type = {a.dtype}")
```

有些数据创建例程不接收形状元组：

```python
# 分配内存并用指定值填充数组、但不接受 shape 作为参数的 NumPy 程序
a = np.arange(4.);              print(f"np.arange(4.):     a = {a}, a shape = {a.shape}, a data type = {a.dtype}")
a = np.random.rand(4);          print(f"np.random.rand(4): a = {a}, a shape = {a.shape}, a data type = {a.dtype}")
```

数值也可以手工指定。

```python
# 分配内存并用用户指定值填充的 NumPy 程序
a = np.array([5,4,3,2]);  print(f"np.array([5,4,3,2]):  a = {a},     a shape = {a.shape}, a data type = {a.dtype}")
a = np.array([5.,4,3,2]); print(f"np.array([5.,4,3,2]): a = {a}, a shape = {a.shape}, a data type = {a.dtype}")
```

这些都创建了一维向量 `a`，含四个元素。`a.shape` 返回各维的长度。这里我们看到 a.shape = `(4,)`，表示含 4 个元素的一维数组。

## 3.4 向量运算
让我们探讨一些使用向量的运算。
<a name="toc_40015_3.4.1"></a>
### 3.4.1 索引
向量的元素可以通过索引和切片访问。NumPy 提供了非常完整的索引与切片能力。这里我们只探讨课程所需的基础内容。更多细节请参阅[切片与索引](https://NumPy.org/doc/stable/reference/arrays.indexing.html)。
**索引（Indexing）**是指按元素在数组中的位置引用数组的*某个元素*。
**切片（Slicing）**是指根据索引取得数组元素的*子集*。
NumPy 的索引从零开始，因此向量 $\mathbf{a}$ 的第 3 个元素是 `a[2]`。

```python
#一维向量的索引操作
a = np.arange(10)
print(a)

#访问一个元素
print(f"a[2].shape: {a[2].shape} a[2]  = {a[2]}, Accessing an element returns a scalar")

# 访问最后一个元素，负索引从末尾开始计数
print(f"a[-1] = {a[-1]}")

#索引必须在该向量的范围内，否则会产生错误
try:
    c = a[10]
except Exception as e:
    print("The error message you'll see is:")
    print(e)
```

<a name="toc_40015_3.4.2"></a>
### 3.4.2 切片
切片用一组三个值（`start:stop:step`）生成索引数组。值的子集写法也是有效的。用例子来说明它的用法最清楚：

```python
#向量切片操作
a = np.arange(10)
print(f"a         = {a}")

#访问 5 个连续元素（起点:终点:步长）
c = a[2:7:1];     print("a[2:7:1] = ", c)

# 访问相隔两个的 3 个元素
c = a[2:7:2];     print("a[2:7:2] = ", c)

# 访问索引 3 及以上的所有元素
c = a[3:];        print("a[3:]    = ", c)

# 访问索引 3 以下的所有元素
c = a[:3];        print("a[:3]    = ", c)

# 访问所有元素
c = a[:];         print("a[:]     = ", c)
```

### 3.4.3 单向量运算
有许多有用的运算只涉及对单个向量的操作。

```python
a = np.array([1,2,3,4])
print(f"a             : {a}")
# 对 a 的元素取负
b = -a 
print(f"b = -a        : {b}")

# 求 a 的所有元素之和，返回一个标量
b = np.sum(a) 
print(f"b = np.sum(a) : {b}")

b = np.mean(a)
print(f"b = np.mean(a): {b}")

b = a**2
print(f"b = a**2      : {b}")
```

### 3.4.4 向量与向量的逐元素运算
NumPy 的大多数算术、逻辑和比较运算同样适用于向量。这些运算符按元素逐一作用。例如
$$ \mathbf{a} + \mathbf{b} = \sum_{i=0}^{n-1} a_i + b_i $$

```python
a = np.array([ 1, 2, 3, 4])
b = np.array([-1,-2, 3, 4])
print(f"Binary operators work element wise: {a + b}")
```

当然，要让它正确工作，两个向量的大小必须相同：

```python
#试试不匹配的向量运算
c = np.array([1, 2])
try:
    d = a + c
except Exception as e:
    print("The error message you'll see is:")
    print(e)
```

### 3.4.6 向量与向量的点积
点积是线性代数和 NumPy 的支柱。本课程中大量使用这一运算，应当充分理解。点积如下所示。

<img src="./图片/C1_W2_Lab04_dot_notrans.gif" width=800>

> **图中文字（中文）**：向量点积（Vector Dot Product） 

点积把两个向量中的值按元素相乘，再把结果求和。
向量点积要求两个向量的维数相同。

让我们在下面实现自己的点积版本：

**用 for 循环**实现一个返回两个向量点积的函数。给定输入 $a$ 和 $b$，函数应返回：
$$ x = \sum_{i=0}^{n-1} a_i b_i $$
假设 `a` 和 `b` 形状相同。

```python
def my_dot(a, b): 
    """计算两个向量的点积
 
    参数：
      a (ndarray (n,)):  输入向量
      b (ndarray (n,)):  与 a 维度相同的输入向量
   
    返回：
      x (scalar):"""
    x=0
    for i in range(a.shape[0]):
        x = x + a[i] * b[i]
    return x
```

```python
# 测试一维
a = np.array([1, 2, 3, 4])
b = np.array([-1, 4, 3, 2])
print(f"my_dot(a, b) = {my_dot(a, b)}")
```

注意，点积应返回一个标量值。

让我们用 `np.dot` 做同样的运算。

```python
# 测试一维
a = np.array([1, 2, 3, 4])
b = np.array([-1, 4, 3, 2])
c = np.dot(a, b)
print(f"NumPy 1-D np.dot(a, b) = {c}, np.dot(a, b).shape = {c.shape} ") 
c = np.dot(b, a)
print(f"NumPy 1-D np.dot(b, a) = {c}, np.dot(a, b).shape = {c.shape} ")
```

上面你会注意到，一维的结果与我们的实现一致。

<a name="toc_40015_3.4.7"></a>
### 3.4.7 速度的必要性：向量与 for 循环
我们使用 NumPy 库，是因为它提升了速度和内存效率。让我们演示一下：

```python
np.random.seed(1)
a = np.random.rand(10000000)  # 非常大的数组
b = np.random.rand(10000000)

tic = time.time()  # 记录开始时间
c = np.dot(a, b)
toc = time.time()  # 记录结束时间

print(f"np.dot(a, b) =  {c:.4f}")
print(f"Vectorized version duration: {1000*(toc-tic):.4f} ms ")

tic = time.time()  # 记录开始时间
c = my_dot(a,b)
toc = time.time()  # 记录结束时间

print(f"my_dot(a, b) =  {c:.4f}")
print(f"loop version duration: {1000*(toc-tic):.4f} ms ")

del(a);del(b)  #把这些大数组从内存中移除
```

可见，在这个例子中向量化带来了大幅提速。这是因为 NumPy 更好地利用了底层硬件中可用的数据并行能力。GPU 和现代 CPU 实现了单指令多数据（SIMD）流水线，可以并行发出多个操作。这在数据集往往非常庞大的机器学习中至关重要。

## 4.4 矩阵运算
让我们探讨一些使用矩阵的运算。

<a name="toc_40015_4.4.1"></a>
### 4.4.1 索引


矩阵有第二个索引。两个索引描述 [行, 列]。访问既可以返回一个元素，也可以返回一行/一列。见下：

```python
#矩阵的向量索引操作
a = np.arange(6).reshape(-1, 2)   #reshape 是创建矩阵的一种便捷方式
print(f"a.shape: {a.shape}, \na= {a}")

#访问一个元素
print(f"\na[2,0].shape:   {a[2, 0].shape}, a[2,0] = {a[2, 0]},     type(a[2,0]) = {type(a[2, 0])} Accessing an element returns a scalar\n")

#访问一行
print(f"a[2].shape:   {a[2].shape}, a[2]   = {a[2]}, type(a[2])   = {type(a[2])}")
```

最后一个例子值得特别提请注意。只指定行来访问矩阵，会返回一个*一维向量*。

**重塑（Reshape）**
上一个例子用 [reshape](https://numpy.org/doc/stable/reference/generated/numpy.reshape.html) 来给数组定形。
`a = np.arange(6).reshape(-1, 2) `
这行代码先创建了含六个元素的*一维向量*，然后用 reshape 命令把该向量重塑为*二维*数组。也可以写成：
`a = np.arange(6).reshape(3, 2) `
以得到同样的 3 行 2 列数组。
参数 -1 告诉该例程：根据数组大小和列数自动计算行数。


<a name="toc_40015_4.4.2"></a>
### 4.4.2 切片
切片用一组三个值（`start:stop:step`）生成索引数组。值的子集写法也是有效的。用例子来说明它的用法最清楚：

```python
#二维向量的切片操作
a = np.arange(20).reshape(-1, 10)
print(f"a = \n{a}")

#访问 5 个连续元素（起点:终点:步长）
print("a[0, 2:7:1] = ", a[0, 2:7:1], ",  a[0, 2:7:1].shape =", a[0, 2:7:1].shape, "a 1-D array")

#在两行中访问 5 个连续元素（起点:终点:步长）
print("a[:, 2:7:1] = \n", a[:, 2:7:1], ",  a[:, 2:7:1].shape =", a[:, 2:7:1].shape, "a 2-D array")

# 访问所有元素
print("a[:,:] = \n", a[:,:], ",  a[:,:].shape =", a[:,:].shape)

# 访问某一行的所有元素（非常常见的用法）
print("a[1,:] = ", a[1,:], ",  a[1,:].shape =", a[1,:].shape, "a 1-D array")
# 等同于
print("a[1]   = ", a[1],   ",  a[1].shape   =", a[1].shape, "a 1-D array")
```
