## 模型函数

<img align="left" src="./图片/C1_W1_L3_S1_model.png"     style=" width:380px; padding: 10px; " >

> **图中文字（中文）**：训练集（Training Set）→ 学习算法（Learning Algorithm）→ f（模型预测 / 假设 / 估计的 y） 特征（feature）x；目标（target）y 如何表示 f ？ f_{w,b}(x) = wx + b 单变量线性回归 如讲课中所述，线性回归的模型函数（一个把 `x` 映射到 `y` 的函数）表示为

$$ f_{w,b}(x^{(i)}) = wx^{(i)} + b \tag{1}$$

上式就是表示直线的方法——不同的 $w$ 和 $b$ 取值会给出图中不同的直线。 <br/> <br/> <br/> <br/> <br/> 

让我们通过下面的代码块建立更直观的理解。先从 $w = 100$ 和 $b = 100$ 开始。

**注意：你可以回到这个单元格调整模型的 w 和 b 参数**

```python
w = 100
b = 100
print(f"w: {w}")
print(f"b: {b}")
```

现在，让我们为你的两个数据点计算 $f_{w,b}(x^{(i)})$ 的值。你可以为每个数据点显式地写出来，即——

对 $x^{(0)}$，`f_wb = w * x[0] + b`

对 $x^{(1)}$，`f_wb = w * x[1] + b`

当数据点很多时，这样写会变得笨拙而重复。因此，你可以像下面 `compute_model_output` 函数那样，用 `for` 循环计算函数输出。
> **注意**：参数说明 `(ndarray (m,))` 描述的是一个形状为 (m,) 的 Numpy n 维数组。`(scalar)` 描述的是一个没有维度的参数，只是一个数值。  
> **注意**：`np.zero(n)` 会返回一个含 $n$ 个元素的一维 numpy 数组   


```python
def compute_model_output(x, w, b):
    """计算线性模型的预测
    参数：
      x (ndarray (m,)): 数据，m 个样本
      w,b (scalar)    : 模型参数
    返回
      y (ndarray (m,)): 目标值"""
    m = x.shape[0]
    f_wb = np.zeros(m)
    for i in range(m):
        f_wb[i] = w * x[i] + b
        
    return f_wb
```

现在让我们调用 `compute_model_output` 函数并绘制输出。

```python
tmp_f_wb = compute_model_output(x_train, w, b,)

# 绘制我们的模型预测
plt.plot(x_train, tmp_f_wb, c='b',label='Our Prediction')

# 绘制数据点
plt.scatter(x_train, y_train, marker='x', c='r',label='Actual Values')

# 设置标题
plt.title("Housing Prices")
# 设置 y 轴标签
plt.ylabel('Price (in 1000s of dollars)')
# 设置 x 轴标签
plt.xlabel('Size (1000 sqft)')
plt.legend()
plt.show()
```

如你所见，把 $w = 100$ 和 $b = 100$ 代入得到的直线并*不*能拟合我们的数据。

### 预测
现在我们有了模型，就可以用它来做最初的预测。让我们预测一套 1200 平方英尺住房的价格。由于 $x$ 的单位是 1000 平方英尺，$x$ 为 1.2。


```python
w = 200                         
b = 100    
x_i = 1.2
cost_1200sqft = w * x_i + b    

print(f"${cost_1200sqft:.0f} thousand dollars")
```
