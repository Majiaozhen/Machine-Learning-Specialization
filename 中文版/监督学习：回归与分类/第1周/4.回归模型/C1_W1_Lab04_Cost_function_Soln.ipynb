## 计算代价
本次作业中“代价（cost）”一词可能有点容易混淆，因为数据本身就是住房的成本。在这里，代价衡量的是我们的模型预测住房目标价格的准确程度。而“价格（price）”一词用于指住房数据。

单变量代价的公式为：
  $$J(w,b) = \frac{1}{2m} \sum\limits_{i = 0}^{m-1} (f_{w,b}(x^{(i)}) - y^{(i)})^2 \tag{1}$$ 
 
其中
  $$f_{w,b}(x^{(i)}) = wx^{(i)} + b \tag{2}$$
  
- $f_{w,b}(x^{(i)})$ 是我们用参数 $w,b$ 对样本 $i$ 的预测。
- $(f_{w,b}(x^{(i)}) -y^{(i)})^2$ 是目标值与预测值之差的平方。
- 这些差值对所有 $m$ 个样本求和，再除以 `2m`，得到代价 $J(w,b)$。
>注意，讲课中求和通常从 1 到 m，而代码中从 0 到 m-1。


下面的代码通过遍历每个样本来计算代价。每次循环中：
- 计算预测值 `f_wb`
- 计算目标值与预测值之差并求平方。
- 把它累加到总代价中。

```python
def compute_cost(x, y, w, b): 
    """计算线性回归的代价函数。
   
    参数：
      x (ndarray (m,)): 数据，m 个样本
      y (ndarray (m,)): 目标值
      w,b (scalar)    : 模型参数
   
    返回
        total_cost (float): 以 w、b 为线性回归参数来拟合 x 和 y 中数据点的代价"""
    # 训练样本数
    m = x.shape[0] 
    
    cost_sum = 0 
    for i in range(m): 
        f_wb = w * x[i] + b   
        cost = (f_wb - y[i]) ** 2  
        cost_sum = cost_sum + cost  
    total_cost = (1 / (2 * m)) * cost_sum  

    return total_cost
```

## 代价函数的直观理解

<img align="left" src="../工作区/图片/C1_W1_Lab02_GoalOfRegression.PNG"    style=" width:380px; padding: 10px;  " />

> **图中文字（中文）**：线性回归的目标：最小化 J(w,b) （左：作为 w、b 的函数；右：作为 w 的函数，图中 w = 0.5） 你的目标是找到一个模型 $f_{w,b}(x) = wx + b$，其参数为 $w,b$，能在给定输入 $x$ 时准确预测住房价值。代价衡量模型在训练数据上的准确程度。

上面的代价公式 (1) 表明，如果能选取 $w$ 和 $b$ 使预测值 $f_{w,b}(x)$ 与目标数据 $y$ 吻合，那么 $(f_{w,b}(x^{(i)}) - y^{(i)})^2 $ 项将为零，代价也被最小化。在这个简单的两点例子中，你是可以做到这一点的！

在上一个实验中，你确定 $b=100$ 给出了最优解，因此我们把 $b$ 设为 100，专注于 $w$。

<br/>
下面，用滑块控件选择使代价最小的 $w$ 值。图更新可能需要几秒钟。

```python
plt_intuition(x_train,y_train)
```

这张图中有几点值得说明。
- 当 $w = 200$ 时代价最小，这与上一个实验的结果一致
- 由于代价公式中目标值与预测值之差取了平方，当 $w$ 过大或过小时代价都会迅速增大。
- 用最小化代价选出的 `w` 和 `b` 得到的直线能完美拟合数据。
