### compute_cost
这是在上一个实验中开发的。这里我们还会用到它。

```python
#计算代价的函数
def compute_cost(x, y, w, b):
   
    m = x.shape[0] 
    cost = 0
    
    for i in range(m):
        f_wb = w * x[i] + b
        cost = cost + (f_wb - y[i])**2
    total_cost = 1 / (2 * m) * cost

    return total_cost
```

## 梯度下降概要
到目前为止，在本课程中你已经建立了一个线性模型来预测 $f_{w,b}(x^{(i)})$：
$$f_{w,b}(x^{(i)}) = wx^{(i)} + b \tag{1}$$
在线性回归中，你利用输入的训练数据，通过最小化预测 $f_{w,b}(x^{(i)})$ 与实际数据 $y^{(i)}$ 之间误差的度量来拟合参数 $w$、$b$。这个度量称为 $cost$（代价），即 $J(w,b)$。训练时，你在所有训练样本 $x^{(i)},y^{(i)}$ 上衡量代价
$$J(w,b) = \frac{1}{2m} \sum\limits_{i = 0}^{m-1} (f_{w,b}(x^{(i)}) - y^{(i)})^2\tag{2}$$ 


讲课中，*梯度下降*被描述为：

$$\begin{align*} \text{repeat}&\text{ until convergence:} \; \lbrace \newline
\;  w &= w -  \alpha \frac{\partial J(w,b)}{\partial w} \tag{3}  \; \newline 
 b &= b -  \alpha \frac{\partial J(w,b)}{\partial b}  \newline \rbrace
\end{align*}$$
其中，参数 $w$、$b$ 同时更新。
梯度定义为：
$$
\begin{align}
\frac{\partial J(w,b)}{\partial w}  &= \frac{1}{m} \sum\limits_{i = 0}^{m-1} (f_{w,b}(x^{(i)}) - y^{(i)})x^{(i)} \tag{4}\\
  \frac{\partial J(w,b)}{\partial b}  &= \frac{1}{m} \sum\limits_{i = 0}^{m-1} (f_{w,b}(x^{(i)}) - y^{(i)}) \tag{5}\\
\end{align}
$$

这里*同时*的意思是：在更新任何参数之前，先把所有参数的偏导数都算出来。

## 实现梯度下降
你将实现针对单个特征的梯度下降算法。你需要三个函数。
- `compute_gradient`，实现上面的公式 (4) 和 (5)
- `compute_cost`，实现上面的公式 (2)（来自上一个实验的代码）
- `gradient_descent`，使用 compute_gradient 和 compute_cost

约定：
- 存放偏导数的 Python 变量按如下模式命名：$\frac{\partial J(w,b)}{\partial b}$ 命名为 `dj_db`。
- w.r.t 是 With Respect To 的缩写，即 $J(wb)$ 关于 $b$ 的偏导数。


<a name="toc_40291_2.3"></a>
### compute_gradient
<a name='ex-01'></a>
`compute_gradient` 实现上面的 (4) 和 (5)，返回 $\frac{\partial J(w,b)}{\partial w}$、$\frac{\partial J(w,b)}{\partial b}$。内嵌的注释说明了这些操作。

```python
def compute_gradient(x, y, w, b): 
    """计算线性回归的梯度
    参数：
      x (ndarray (m,)): 数据，m 个样本
      y (ndarray (m,)): 目标值
      w,b (scalar)    : 模型参数
    返回
      dj_dw (scalar): 代价对参数 w 的梯度
      dj_db (scalar): 代价对参数 b 的梯度"""
    
    # 训练样本数
    m = x.shape[0]    
    dj_dw = 0
    dj_db = 0
    
    for i in range(m):  
        f_wb = w * x[i] + b 
        dj_dw_i = (f_wb - y[i]) * x[i] 
        dj_db_i = f_wb - y[i] 
        dj_db += dj_db_i
        dj_dw += dj_dw_i 
    dj_dw = dj_dw / m 
    dj_db = dj_db / m 
        
    return dj_dw, dj_db
```

左图固定了 $b=100$。梯度下降将同时利用 $\frac{\partial J(w,b)}{\partial w}$ 和 $\frac{\partial J(w,b)}{\partial b}$ 来更新参数。右侧的“箭头图（quiver plot）”提供了一种同时观察两个参数梯度的手段。箭头大小反映该点处梯度的幅值。箭头的方向和斜率反映该点处 $\frac{\partial J(w,b)}{\partial w}$ 与 $\frac{\partial J(w,b)}{\partial b}$ 的比值。
注意，梯度指向的是*远离*最小值的方向。回顾上面的公式 (3)：缩放后的梯度是从 $w$ 或 $b$ 的当前值中*减去*的。这会把参数朝降低代价的方向移动。

###  梯度下降
既然梯度可以计算了，上面公式 (3) 所描述的梯度下降就可以在下面用 `gradient_descent` 实现。实现的细节在注释中说明。下面，你将用这个函数在训练数据上找出 $w$ 和 $b$ 的最优值。

```python
def gradient_descent(x, y, w_in, b_in, alpha, num_iters, cost_function, gradient_function): 
    """执行梯度下降以拟合 w、b。以学习率 alpha 走
    num_iters 个梯度步来更新 w、b
   
    参数：
      x (ndarray (m,))  : 数据，m 个样本
      y (ndarray (m,))  : 目标值
      w_in,b_in (scalar): 模型参数的初始值
      alpha (float):     学习率
      num_iters (int):   运行梯度下降的迭代次数
      cost_function:     用于计算代价的函数
      gradient_function: 用于计算梯度的函数
     
    返回：
      w (scalar): 运行梯度下降后更新得到的参数值
      b (scalar): 运行梯度下降后更新得到的参数值
      J_history (List): 代价取值的历史记录
      p_history (list): 参数 [w,b] 的历史记录"""
    
    w = copy.deepcopy(w_in) # 避免修改全局变量 w_in
    # 用于存放代价 J 以及每次迭代的 w，主要供后续绘图使用
    J_history = []
    p_history = []
    b = b_in
    w = w_in
    
    for i in range(num_iters):
        # 用 gradient_function 计算梯度并更新参数
        dj_dw, dj_db = gradient_function(x, y, w , b)     

        # 用上面的公式 (3) 更新参数
        b = b - alpha * dj_db                            
        w = w - alpha * dj_dw                            

        # 每次迭代保存代价 J
        if i<100000:      # 防止资源耗尽
            J_history.append( cost_function(x, y, w , b))
            p_history.append([w,b])
        # 每 10 次迭代打印一次代价；若迭代次数少于 10，则打印所有迭代
        if i% math.ceil(num_iters/10) == 0:
            print(f"Iteration {i:4}: Cost {J_history[-1]:0.2e} ",
                  f"dj_dw: {dj_dw: 0.3e}, dj_db: {dj_db: 0.3e}  ",
                  f"w: {w: 0.3e}, b:{b: 0.5e}")
 
    return w, b, J_history, p_history #返回 w 以及用于绘图的 J、w 历史
```

```python
# 初始化参数
w_init = 0
b_init = 0
# 一些梯度下降设置
iterations = 10000
tmp_alpha = 1.0e-2
# 运行梯度下降
w_final, b_final, J_hist, p_hist = gradient_descent(x_train ,y_train, w_init, b_init, tmp_alpha, 
                                                    iterations, compute_cost, compute_gradient)
print(f"(w,b) found by gradient descent: ({w_final:8.4f},{b_final:8.4f})")
```

### 代价随梯度下降迭代次数的变化
把代价随迭代次数作图，是衡量梯度下降进展的有用方法。在成功的运行中，代价应始终下降。代价在最初变化非常快，因此把初始下降段和最终下降段用不同的坐标尺度来绘制会更有用。在下面的图中，注意坐标轴上代价的尺度以及迭代步数。

```python
# 绘制代价随迭代次数的变化
fig, (ax1, ax2) = plt.subplots(1, 2, constrained_layout=True, figsize=(12,4))
ax1.plot(J_hist[:100])
ax2.plot(1000 + np.arange(len(J_hist[1000:])), J_hist[1000:])
ax1.set_title("Cost vs. iteration(start)");  ax2.set_title("Cost vs. iteration (end)")
ax1.set_ylabel('Cost')            ;  ax2.set_ylabel('Cost') 
ax1.set_xlabel('iteration step')  ;  ax2.set_xlabel('iteration step') 
plt.show()
```

### 增大学习率

<figure>
 <img align="left", src="../工作区/图片/C1_W1_Lab03_alpha_too_big.PNG"   style="width:340px;height:240px;" >

> **图中文字（中文）**：若 α 过小，梯度下降可能很慢。 若 α 过大，梯度下降可能 - 越过最小值，永远到达不了最小值 - 无法收敛、发散
</figure>
讲课中讨论了公式 (3) 中学习率 $\alpha$ 的合适取值。$\alpha$ 越大，梯度下降收敛到解就越快。但如果它过大，梯度下降就会发散。上面是一个收敛良好的解的示例。

让我们试着增大 $\alpha$ 的值，看看会发生什么：

```python
# 初始化参数
w_init = 0
b_init = 0
# 把 alpha 设为一个很大的值
iterations = 10
tmp_alpha = 8.0e-1
# 运行梯度下降
w_final, b_final, J_hist, p_hist = gradient_descent(x_train ,y_train, w_init, b_init, tmp_alpha, 
                                                    iterations, compute_cost, compute_gradient)
```

上面，$w$ 和 $b$ 在正负之间来回震荡，且绝对值随每次迭代增大。此外，每次迭代 $\frac{\partial J(w,b)}{\partial w}$ 都改变符号，代价不降反升。这清楚地表明*学习率过大*，解正在发散。
让我们用图来可视化这一点。

```python
plt_divergence(p_hist, J_hist,x_train, y_train)
plt.show()
```

上面，左图显示了梯度下降前几步中 $w$ 的变化过程。$w$ 在正负之间振荡，代价迅速增大。梯度下降是同时对 $w$ 和 $b$ 作用的，所以要看到完整的情况，需要看右边的三维图。
