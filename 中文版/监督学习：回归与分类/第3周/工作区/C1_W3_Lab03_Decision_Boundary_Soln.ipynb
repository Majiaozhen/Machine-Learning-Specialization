## 数据集

假设你有下面这个训练数据集
- 输入变量 `X` 是一个 numpy 数组，包含 6 个训练样本，每个样本有两个特征

```python
X = np.array([[0.5, 1.5], [1,1], [1.5, 0.5], [3, 0.5], [2, 2], [1, 2.5]])
y = np.array([0, 0, 0, 1, 1, 1]).reshape(-1,1) 
```

让我们用一个辅助函数来绘制这些数据。标签为 $y=1$ 的数据点显示为红色叉号，标签为 $y=0$ 的数据点显示为蓝色圆圈。

```python
fig,ax = plt.subplots(1,1,figsize=(4,4))
plot_data(X, y, ax)

ax.axis([0, 4, 0, 3.5])
ax.set_ylabel('$x_1$')
ax.set_xlabel('$x_0$')
plt.show()
```

## 逻辑回归模型


* 假设你想在这份数据上训练一个逻辑回归模型，其形式为  

  $f(x) = g(w_0x_0+w_1x_1 + b)$
  
  其中 $g(z) = \\frac{1}{1+e^{-z}}$，即 sigmoid 函数


* 假设你训练完模型，得到的参数是 $b = -3, w_0 = 1, w_1 = 1$。也就是说，

  $f(x) = g(x_0+x_1-3)$

  （你将在后续课程中学习如何把这些参数拟合到数据上）
  
  
让我们通过绘制决策边界，来理解这个训练好的模型在预测什么

### 逻辑回归与决策边界回顾

* 回忆一下，逻辑回归的模型表示为

  $$f_{\\mathbf{w},b}(\\mathbf{x}^{(i)}) = g(\\mathbf{w} \\cdot \\mathbf{x}^{(i)} + b) \\tag{1}$$

  其中 $g(z)$ 称为 sigmoid 函数，它把所有输入值映射到 0 和 1 之间：

  $g(z) = \\frac{1}{1+e^{-z}}\\tag{2}$
  而 $\\mathbf{w} \\cdot \\mathbf{x}$ 是向量点积：
  
  $$\\mathbf{w} \\cdot \\mathbf{x} = w_0 x_0 + w_1 x_1$$
  
  
 * 我们把模型的输出（$f_{\\mathbf{w},b}(x)$）解释为在给定 $x$、并以 $w$ 和 $b$ 为参数时 $y=1$ 的概率。
* 因此，要从逻辑回归模型得到最终预测（$y=0$ 或 $y=1$），我们可以使用下面的启发式规则——

  若 $f_{\\mathbf{w},b}(x) >= 0.5$，预测 $y=1$
  
  若 $f_{\\mathbf{w},b}(x) < 0.5$，预测 $y=0$
  
  
* 让我们画出 sigmoid 函数，看看 $g(z) >= 0.5$ 的位置

```python
# 绘制 -10 到 10 范围内 sigmoid(z) 的曲线
z = np.arange(-10,11)

fig,ax = plt.subplots(1,1,figsize=(5,3))
# 绘制 z 与 sigmoid(z) 的关系
ax.plot(z, sigmoid(z), c="b")

ax.set_title("Sigmoid function")
ax.set_ylabel('sigmoid(z)')
ax.set_xlabel('z')
draw_vthresh(ax,0)
```

* 可以看到，当 $z >=0$ 时 $g(z) >= 0.5$

* 对于逻辑回归模型，$z = \\mathbf{w} \\cdot \\mathbf{x} + b$。因此，

  若 $\\mathbf{w} \\cdot \\mathbf{x} + b >= 0$，模型预测 $y=1$
  
  若 $\\mathbf{w} \\cdot \\mathbf{x} + b < 0$，模型预测 $y=0$
  
  
  
### 绘制决策边界

现在，让我们回到例子中，理解逻辑回归模型是如何做预测的。

* 我们的逻辑回归模型形式为

  $f(x) = g(-3 + x_0+x_1)$


* 根据上面学到的内容，你可以看出当 $-3 + x_0+x_1 >= 0$ 时该模型预测 $y=1$

让我们看看这在图形上是什么样子。我们先画 $-3 + x_0+x_1 = 0$，它等价于 $x_1 = 3 - x_0$。


```python
# 在 0 到 6 之间取值
x0 = np.arange(0,6)

x1 = 3 - x0
fig,ax = plt.subplots(1,1,figsize=(5,4))
# 绘制决策边界
ax.plot(x0,x1, c="b")
ax.axis([0, 4, 0, 3.5])

# 填充直线下方的区域
ax.fill_between(x0,x1, alpha=0.2)

# 绘制原始数据
plot_data(X,y,ax)
ax.set_ylabel(r'$x_1$')
ax.set_xlabel(r'$x_0$')
plt.show()
```

* 在上图中，蓝线表示直线 $x_0 + x_1 - 3 = 0$，它应当与 x1 轴相交于 3（若设 $x_1$ = 3，则 $x_0$ = 0），与 x0 轴相交于 3（若设 $x_1$ = 0，则 $x_0$ = 3）。


* 阴影区域表示 $-3 + x_0+x_1 < 0$。直线之上的区域是 $-3 + x_0+x_1 > 0$。


* 阴影区域内（直线下方）的任意点都被分类为 $y=0$。直线之上或直线上的任意点都被分类为 $y=1$。这条直线称为“决策边界”。

正如课上所见，通过使用更高阶的多项式项（例如 $f(x) = g( x_0^2 + x_1 -1)$），我们可以得到更复杂的非线性边界。
