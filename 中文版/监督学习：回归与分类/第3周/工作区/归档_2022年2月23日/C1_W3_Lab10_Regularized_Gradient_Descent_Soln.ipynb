## 带正则化的梯度

带正则化代价函数的梯度，即代价对参数 $w$ 和 $b$ 的偏导数：
$$\\begin{align*}
\\frac{\\partial J(\\mathbf{w})}{\\partial b} &= \\frac{1}{m}  \\sum_{i=0}^{m-1} (f_{\\mathbf{w},b}(\\mathbf{x}^{(i)}) - y^{(i)})  \\tag {1} \\\\
\\frac{\\partial J(\\mathbf{w},b)}{\\partial w_j} &= \\left( \\frac{1}{m}  \\sum_{i=0}^{m-1} (f_{\\mathbf{w},b}(\\mathbf{x}^{(i)}) - y^{(i)}) x_j^{(i)} \\right) +  \\frac{\\lambda}{m} w_j  \\quad\\, \\mbox{for $j=0...(n-1)$} \\tag {2}
\\end{align*}$$


你将实现一个名为 `compute_gradient_reg` 的函数，返回 $\\frac{\\partial J(\\mathbf{w},b)}{\\partial w},\\frac{\\partial J(\\mathbf{w},b)}{\\partial b}$。

请完成 `compute_gradient_reg` 函数：

- 完全按照你在之前实验的 `compute_gradient` 函数中的做法，计算每个元素 `dj_dw` 和 `dj_db` 的梯度：
    - 初始化用于累加 `dj_dw` 和 `dj_db` 的变量
    - 遍历所有样本
        - 计算该样本的误差 $g(\\mathbf{x}^{(i)T}\\mathbf{w} + b) - \\mathbf{y}^{(i)}$
        - 把误差加到 `dj_db`（上面的公式 1）
        - 对该样本中的每个输入值 $x_{j}^{(i)}$
            - 把误差乘以输入 $x_{j}^{(i)}$，加到 `dj_dw` 对应的元素上。（上面的公式 2）
     - 把 `dj_db` 和 `dj_dw` 除以样本总数（m）
- 现在计算正则化项
    - 遍历所有 $w$
        - 把 $\\frac{\\lambda}{m} * w_j$ 加到 `dj_dw` 对应的元素上

在做这件事时请记住，变量 X 和 y 不是标量值，而是形状分别为 ($m, n$) 和 ($𝑚$,1 ) 的矩阵，其中 $𝑛$ 是特征数，$𝑚$ 是训练样本数。


````python
def compute_gradient_reg(X, y, w, b, lambda_ = 1): 
    """计算线性回归的梯度
 
    参数：
      X : (array_like 形状 (m,n)) 变量，例如房屋面积
      y : (array_like 形状 (m,1)) 实际值
      w : (array_like 形状 (n,1)) 模型参数取值
      b : (scalar)                模型参数取值
      lambda_ : (scalar,float)     正则化常数
    返回
      dj_dw: (array_like 形状 (n,1)) 代价对参数 w 的梯度。
      dj_db: (scalar)                代价对参数 b 的梯度。"""
    m,n = X.shape
    dj_dw = np.zeros((n,1))
    dj_db = 0.
    err  = 0.

    for i in range(m):
        err = sigmoid(X[i] @ w + b)  - y[i]    
        for j in range(n):
            dj_dw[j] = dj_dw[j] + err * X[i][j]
        dj_db = dj_db + err
    dj_dw = dj_dw/m
    dj_db = dj_db/m
    
    for j in range(n):
        dj_dw[j] = dj_dw[j] + (lambda_/m) * w[j]

        
    return dj_db[0],dj_dw  #对 dj_db 取索引以返回标量值
````

运行下面的单元格，检查你对 `compute_gradient_reg` 函数的实现。

````python
np.random.seed(1)
X_tmp = np.random.rand(5,3)
y_tmp = np.array([0,1,0,1,0]).reshape(-1,1)
initial_w  = np.random.rand(X_tmp.shape[1]).reshape(-1,1)
initial_b = 0.5
lambda_ = 1
dj_db, dj_dw =  compute_gradient_reg(X_tmp, y_tmp, initial_w, initial_b, lambda_)

print(f"dj_db: {dj_db}", )
print(f"Regularized dj_dw:\n {dj_dw.tolist()}", )
````
