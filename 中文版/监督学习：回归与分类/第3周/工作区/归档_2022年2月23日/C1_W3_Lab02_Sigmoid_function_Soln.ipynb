## Sigmoid 函数的公式

sigmoid 函数的公式如下——

$g(z) = \\frac{1}{1+e^{-z}}\\tag{1}$

在逻辑回归中，z（sigmoid 函数的输入）是线性回归模型的输出。
- 也就是说，$z$ 并不总是一个数字，也可能是数字数组。
- 如果输入是数字数组，我们要对输入数组中的每个值应用 sigmoid 函数。

让我们用 Python 实现它。

NumPy 有一个名为 [`exp()`](https://numpy.org/doc/stable/reference/generated/numpy.exp.html) 的函数，它提供了一种便捷的方法来计算输入数组（`z`）中所有元素的指数（$e^{z}$）。

它也可以接受单个数字作为输入，如下所示。

```python
# 输入是数组。
input_array = np.array([1,2,3])
exp_array = np.exp(input_array)

print("Input to exp:", input_array)
print("Output of exp:", exp_array)

# 输入是单个数字
input_val = 1  
exp_val = np.exp(input_val)

print("Input to exp:", input_val)
print("Output of exp:", exp_val)
```

`sigmoid` 函数在 Python 中的实现如下面的单元格所示。

```python
def sigmoid(z):

    g = 1/(1+np.exp(-z))
   
    return g
```

让我们看看这个函数在 `z` 取不同值时的输出

```python
# 生成一个在 -10 到 10 之间均匀取值的数组
z = np.arange(-10,11)

# 用上面实现的函数计算 sigmoid 值
y = sigmoid(z)

# 把两个数组并排漂亮打印出来的代码
np.set_printoptions(precision=3) 
print("Input (z), Output (sigmoid(z))")
print(np.c_[z, y])
```

左列的值是 `z` 的取值，右列的值是 `sigmoid(z)`。可以看到，sigmoid 的输入取值范围为 -10 到 10，输出取值范围都应在 0 到 1 之间。

现在，让我们用 `matplotlib` 库试着把这个函数画出来。

```python
import matplotlib.pyplot as plt

fig,ax = plt.subplots(1,1,figsize=(5,3))
# 绘制 z 与 sigmoid(z) 的关系
ax.plot(z, y, c="b")

ax.set_title("Sigmoid function")
ax.set_ylabel('sigmoid(z)')
ax.set_xlabel('z')
draw_vthresh(ax,0)
```

可以看到，当 `z` 取很大的负值时 sigmoid 函数趋近于 `0`，当 `z` 取很大的正值时趋近于 `1`。


## 逻辑回归
<img align="left" src="./图片/C1_W3_LogisticRegression_right.png"     style=" width:300px; padding: 10px; " >

> **图中文字（中文）**：f_{w,b}(x) = 1/(1 + e^(−(w·x + b))) 逻辑回归 逻辑回归模型把 sigmoid 应用到我们熟悉的线性回归模型上，如下所示：

$$ f_{\\mathbf{w},b}(\\mathbf{x}) = g(\\mathbf{w} \\cdot \\mathbf{x} + b ) \\tag{2} $$ 

其中

  $g(z) = \\frac{1}{1+e^{-z}}\\tag{3}$
  
  且
  
  $$ z = \\mathbf{w} \\cdot \\mathbf{x} + b \\tag{4} $$
