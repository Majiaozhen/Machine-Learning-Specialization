## 逻辑回归能用平方误差吗？
<img align="left" src="./图片/C1_W3_SqErrorVsLogistic.png"     style=" width:400px; padding: 10px; " >

> **图中文字（中文）**：平方误差代价：J(w,b) = (1/2m)Σ(f_{w,b}(x) − y)² —— 对线性回归是凸的 逻辑回归的损失函数（单个训练样本的损失）： L(f_{w,b}(x), y) = −log(f_{w,b}(x))，若 y = 1；−log(1 − f_{w,b}(x))，若 y = 0 整个训练集的平均损失 → 凸函数 在线性回归中，课程使用平方误差代价来衡量模型对训练数据的预测效果。  
回忆一下，**线性**回归中使用的是**平方误差代价函数**：
单变量平方误差代价的方程为：
  $$J(w,b) = \\frac{1}{2m} \\sum\\limits_{i = 0}^{m-1} (f_{w,b}(x^{(i)}) - y^{(i)})^2 \\tag{1}$$ 
 
其中
  $$f_{w,b}(x^{(i)}) = wx^{(i)} + b \\tag{2}$$


回忆一下，平方误差代价有一个很好的性质：沿代价的导数方向走可以到达最小值。

```python
soup_bowl()
```

这个代价函数对线性回归效果很好，自然也会考虑把它用于逻辑回归。然而，正如上面的幻灯片所指出的，$f_{wb}(x)$ 现在含有非线性成分，即 sigmoid 函数：$f_{w,b}(x^{(i)}) = sigmoid(wx^{(i)} + b )$。让我们在之前实验的例子上试用平方误差代价，此时已包含 sigmoid。

下面是我们的训练数据：

```python
x_train = np.array([0., 1, 2, 3, 4, 5],dtype=np.longdouble)
y_train = np.array([0,  0, 0, 1, 1, 1],dtype=np.longdouble)
plt_simple_example(x_train, y_train)
```

现在，让我们用*平方误差代价*得到代价的曲面图：
  $$J(w,b) = \\frac{1}{2m} \\sum\\limits_{i = 0}^{m-1} (f_{w,b}(x^{(i)}) - y^{(i)})^2 $$ 
 
其中
  $$f_{w,b}(x^{(i)}) = sigmoid(wx^{(i)} + b )$$


```python
plt.close('all')
plt_logistic_squared_error(x_train,y_train)
plt.show()
```

虽然这画出了一幅相当有意思的图，但上面的曲面远没有线性回归的“汤碗”那样平滑！

逻辑回归需要一个更适合其非线性特性的代价函数。这要从损失函数说起，下面将进行介绍。

## 逻辑损失函数
<img align="left" src="./图片/C1_W3_LogisticLoss_a.png"     style=" width:250px; padding: 2px; " >

> **图中文字（中文）**：逻辑回归的损失函数 L(f_{w,b}(x), y) = −log(f_{w,b}(x))，若 y = 1 L(f_{w,b}(x), y) = −log(1 − f_{w,b}(x))，若 y = 0 当 f_{w,b}(x) 预测的 y 接近真实标签时，损失最低
<img align="left" src="./图片/C1_W3_LogisticLoss_b.png"     style=" width:250px; padding: 2px; " >

> **图中文字（中文）**：逻辑回归的损失函数 当 y = 1 时，若 f_{w,b}(x) → 1 则损失 → 0；若 f_{w,b}(x) → 0 则损失 → ∞ 当 y = 0 时，若 f_{w,b}(x) → 0 则损失 → 0；若 f_{w,b}(x) → 1 则损失 → ∞ 预测 f_{w,b}(x) 离目标 y 越远，损失越大
<img align="left" src="./图片/C1_W3_LogisticLoss_c.png"     style=" width:250px; padding: 2px; " >

> **图中文字（中文）**：代价函数 J(w,b) = (1/m)Σ L(f_{w,b}(x(i)), y(i)) 整个训练集上的平均损失 


逻辑回归使用的损失函数更适合目标为 0 或 1（而非任意数值）的分类任务。

>**定义提醒：**  
**损失（Loss）** 衡量单个样本与其目标值之间的差异，而  
**代价（Cost）** 衡量整个训练集上的损失


其定义如下：
* $loss(f_{\\mathbf{w},b}(\\mathbf{x}^{(i)}), y^{(i)})$ 是单个数据点的代价，即：

\\begin{equation}
  loss(f_{\\mathbf{w},b}(\\mathbf{x}^{(i)}), y^{(i)}) = \\begin{cases}
    - \\log\\left(f_{\\mathbf{w},b}\\left( \\mathbf{x}^{(i)} \\right) \\right) & \\text{if $y^{(i)}=1$}\\\\
    \\log \\left( 1 - f_{\\mathbf{w},b}\\left( \\mathbf{x}^{(i)} \\right) \\right) & \\text{if $y^{(i)}=0$}
  \\end{cases}
\\end{equation}


*  $f_{\\mathbf{w},b}(\\mathbf{x}^{(i)})$ 是模型的预测，而 $y^{(i)}$ 是目标值。

*  $f_{\\mathbf{w},b}(\\mathbf{x}^{(i)}) = g(\\mathbf{w} \\cdot\\mathbf{x}^{(i)}+b)$，其中函数 $g$ 是 sigmoid 函数。

这个损失函数的特点在于它使用了两条分开的曲线：一条用于目标为零（$y=0$）的情形，另一条用于目标为一（$y=1$）的情形。两条曲线合在一起，提供了作为损失函数所需的行为：当预测与目标一致时取值为零，当预测偏离目标时取值迅速增大。请看下面的曲线：

```python
plt_two_logistic_loss_curves()
```

两条曲线合在一起，与平方误差损失的抛物线相似。注意，横轴是 $f_{\\mathbf{w},b}$，即 sigmoid 的输出。sigmoid 的输出严格位于 0 和 1 之间。

上面的损失函数可以改写成更便于在实现代价的程序中使用的形式。
    $$loss(f_{\\mathbf{w},b}(\\mathbf{x}^{(i)}), y^{(i)}) = (-y^{(i)} \\log\\left(f_{\\mathbf{w},b}\\left( \\mathbf{x}^{(i)} \\right) \\right) - \\left( 1 - y^{(i)}\\right) \\log \\left( 1 - f_{\\mathbf{w},b}\\left( \\mathbf{x}^{(i)} \\right) \\right)$$
  
这是一个看起来相当吓人的方程。但考虑到 $y^{(i)}$ 只能取两个值 0 和 1，它就不那么可怕了。于是可以把方程分成两部分来看：
当 $ y^{(i)} = 0$ 时，左边一项被消去：
$$
\\begin{align}
loss(f_{\\mathbf{w},b}(\\mathbf{x}^{(i)}), 0) &= (-(0) \\log\\left(f_{\\mathbf{w},b}\\left( \\mathbf{x}^{(i)} \\right) \\right) - \\left( 1 - 0\\right) \\log \\left( 1 - f_{\\mathbf{w},b}\\left( \\mathbf{x}^{(i)} \\right) \\right) \\\\
&= -\\log \\left( 1 - f_{\\mathbf{w},b}\\left( \\mathbf{x}^{(i)} \\right) \\right)
\\end{align}
$$
当 $ y^{(i)} = 1$ 时，右边一项被消去：
$$
\\begin{align}
  loss(f_{\\mathbf{w},b}(\\mathbf{x}^{(i)}), 1) &=  (-(1) \\log\\left(f_{\\mathbf{w},b}\\left( \\mathbf{x}^{(i)} \\right) \\right) - \\left( 1 - 1\\right) \\log \\left( 1 - f_{\\mathbf{w},b}\\left( \\mathbf{x}^{(i)} \\right) \\right)\\\\
  &=  -\\log\\left(f_{\\mathbf{w},b}\\left( \\mathbf{x}^{(i)} \\right) \\right)
\\end{align}
$$

好，有了这个新的逻辑损失函数，就可以构建一个汇总所有样本损失的代价函数了。这将是下一个实验的主题。现在，让我们先看看上面所考虑的简单例子中代价随参数变化的曲线：

```python
plt.close('all')
cst = plt_logistic_cost(x_train,y_train)
```

这条曲线非常适合梯度下降！它没有平台、没有局部最小值，也没有不连续处。注意，它不像平方误差那样是一个碗形。图中同时绘制了代价和代价的对数，以说明当代价很小时曲线仍有斜率并继续下降。提醒：你可以用鼠标旋转上面的图。
