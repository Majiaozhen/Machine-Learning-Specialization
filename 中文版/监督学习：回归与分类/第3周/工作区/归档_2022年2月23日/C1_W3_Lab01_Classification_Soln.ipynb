## 分类问题

> **图中文字（中文）**：分类问题 这是垃圾邮件吗？ 是 / 否 这笔交易是欺诈吗？ 是 / 否 这个肿瘤是恶性的吗？ 是 / 否 y 只能取两个值之一：False / 0 —— “负类（negative class）”；True / 1 —— “正类（positive class）” 二元分类（binary classification） 分类问题的例子有：判断邮件是垃圾邮件还是非垃圾邮件，或者判断肿瘤是恶性还是良性。特别地，这些都是*二元*分类的例子，即只有两种可能的结果。结果可以用“正”/“负”这样的成对方式描述，例如“是”/“否”、“真”/“假”或“1”/“0”。

分类数据集的图通常用符号来表示样本的结果。在下面的图中，用 'X' 表示正类取值，用 'O' 表示负类结果。

```python
x_train = np.array([0., 1, 2, 3, 4, 5])
y_train = np.array([0,  0, 0, 1, 1, 1])
X_train2 = np.array([[0.5, 1.5], [1,1], [1.5, 0.5], [3, 0.5], [2, 2], [1, 2.5]])
y_train2 = np.array([0, 0, 0, 1, 1, 1])
```

```python
pos = y_train == 1
neg = y_train == 0

fig,ax = plt.subplots(1,2,figsize=(8,3))
#图 1，单变量
ax[0].scatter(x_train[pos], y_train[pos], marker='x', s=80, c = 'red', label="y=1")
ax[0].scatter(x_train[neg], y_train[neg], marker='o', s=100, label="y=0", facecolors='none', edgecolors=dlblue,lw=3)

ax[0].set_ylim(-0.08,1.1)
ax[0].set_ylabel('y', fontsize=12)
ax[0].set_xlabel('x', fontsize=12)
ax[0].set_title('one variable plot')
ax[0].legend()

#图 2，双变量
plot_data(X_train2, y_train2, ax[1])
ax[1].axis([0, 4, 0, 4])
ax[1].set_ylabel('$x_1$', fontsize=12)
ax[1].set_xlabel('$x_0$', fontsize=12)
ax[1].set_title('two variable plot')
ax[1].legend()
plt.tight_layout()
plt.show()
```

注意上面图中的几点：
- 在单变量图中，正结果同时用红色的 'X' 和 y=1 表示。负结果是蓝色的 'O'，位于 y=0。
   - 回忆一下，在线性回归中，y 不会被限制为两个取值，而可以是任意值。
- 在双变量图中，无法显示 y 轴。正结果同样用红色 'X' 表示，负结果用蓝色 'O' 符号表示。
    - 回忆一下，在多变量线性回归中，y 不会被限制为两个取值，类似的图会是三维的。
