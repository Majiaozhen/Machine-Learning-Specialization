## 循环版本

带正则化代价函数的梯度，即代价对参数 $w$ 和 $b$ 的偏导数：

$$\\frac{\\partial J(\\mathbf{w})}{\\partial b} = \\frac{1}{m}  \\sum_{i=0}^{m-1} (f_{\\mathbf{w},b}(\\mathbf{x}^{(i)}) - y^{(i)})  \\tag {1}$$

$$\\frac{\\partial J(\\mathbf{w},b)}{\\partial w_j} = \\left( \\frac{1}{m}  \\sum_{i=0}^{m-1} (f_{\\mathbf{w},b}(\\mathbf{x}^{(i)}) - y^{(i)}) x_j^{(i)} \\right) + \\lambda w_j  \\quad\\, \\mbox{for $j=0...(n-1)$} \\tag {2}$$


你将实现一个名为 `compute_gradient_reg` 的函数，返回 $\\frac{\\partial J(\\mathbf{w},b)}{\\partial w},\\frac{\\partial J(\\mathbf{w},b)}{\\partial b}$。

请完成 `compute_gradient_reg` 函数：

- 完全按照你在之前实验的 `compute_gradient` 函数中的做法，计算每个元素 `dJdw` 和 `dJdb` 的梯度：
    - 初始化用于累加 `dJdw` 和 `dJdb` 的变量
    - 遍历所有样本
        - 计算该样本的误差 $g(\\mathbf{x}^{(i)T}\\mathbf{w} + b) - \\mathbf{y}^{(i)}$
        - 把误差加到 `dJdb`（上面的公式 1）
        - 对该样本中的每个输入值 $x_{j}^{(i)}$
            - 把误差乘以输入 $x_{j}^{(i)}$，加到 `dJdw` 对应的元素上。（上面的公式 2）
     - 把 `dJdb` 和 `dJdw` 除以样本总数（m）
- 现在计算正则化项
    - 遍历所有 $w$
        - 把 $\\lambda * w_j$ 加到 `dJdw` 对应的元素上

在做这件事时请记住，变量 X 和 y 不是标量值，而是形状分别为 ($m, n$) 和 ($𝑚$,1 ) 的矩阵，其中 $𝑛$ 是特征数，$𝑚$ 是训练样本数。
