## 逻辑回归模型


* 假设你想在这份数据上训练一个逻辑回归模型，其形式为  

  $f(x) = g(w_0x_0+w_1x_1 + b)$
  
  其中 $g(z) = \\frac{1}{1+e^{-z}}$，即 sigmoid 函数


* 假设你训练完模型，得到的参数是 $b = -3, w_0 = 1, w_1 = 1$。也就是说，

  $f(x) = g(x_0+x_1-3)$

  （你将在后续课程中学习如何把这些参数拟合到数据上）
  
  
让我们通过绘制决策边界，来理解这个训练好的模型在预测什么

### 逻辑回归与决策边界回顾

* 回忆一下，逻辑回归的模型表示为

  $f_{\\mathbf{w},b}(x) = g(x^Tw)$

  其中

  $g(z) = \\frac{1}{1+e^{-z}}$
  
  $g(z)$ 称为 sigmoid 函数，它把所有输入值映射到 0 和 1 之间。
  

* 我们把模型的输出（$f_{\\mathbf{w},b}(x)$）解释为在给定 $x$、并以 $b$ 和 $w$ 为参数时 $y=1$ 的概率。
* 因此，要从逻辑回归模型得到最终预测（$y=0$ 或 $y=1$），我们可以使用下面的启发式规则——

  若 $f_{\\mathbf{w},b}(x) >= 0.5$，预测 $y=1$
  
  若 $f_{\\mathbf{w},b}(x) < 0.5$，预测 $y=0$
  
  
* 由于 $f_{\\mathbf{w},b}(x) = g(x^Tw)$，让我们画出 sigmoid 函数，看看 $g(z) >= 0.5$ 的位置

我们已经为你实现了 `sigmoid` 函数，你只需导入并使用它，如下面的代码块所示。

```python
import matplotlib.pyplot as plt
import numpy as np
from lab_utils import sigmoid


# 绘制 -20 到 20 范围内 sigmoid(z) 的曲线
z = np.arange(-20,20)
plt.plot(z, sigmoid(z), c="b")

# 绘制并标注 sigmoid(z) = 0.5 的点
plt.plot(0, sigmoid(0), 'ro')
plt.text(0, sigmoid(0), '(x={}, y={})'.format(0, sigmoid(0)))
```

* 可以看到，当 $z >=0$ 时 $g(z) >= 0.5$

* 对于逻辑回归模型，$z = x^Tw$。因此，

  若 $x^Tw >= 0$，模型预测 $y=1$
  
  若 $x^Tw < 0$，模型预测 $y=0$
  
  
  
### 绘制决策边界

现在，让我们回到例子中，理解逻辑回归模型是如何做预测的。

* 我们的逻辑回归模型形式为

  $f(x) = g(-3 + x_0+x_1)$


* 根据上面学到的内容，你可以看出当 $-3 + x_0+x_1 >= 0$ 时该模型预测 $y=1$

让我们看看这在图形上是什么样子。我们先画 $-3 + x_0+x_1 = 0$，它等价于 $x_1 = 3 - x_0$。

**练习**

在下面的代码块中写出 `x1 = 3-x0` 以完成它

```python
# 在 0 到 6 之间取值
x0 = np.arange(0,6)

### 在此处编写代码 ###
### 解答开始 ###
x1 = 3 - x0
### 代码编写结束 ###
### 解答结束 ###
# 绘制决策边界
plt.plot(x0,x1, c="b")
plt.axis([0, 6, 0, 6])

# 填充直线下方的区域
plt.fill_between(x0,x1, alpha=0.2)

# 绘制原始数据
plot_data(X,y)
# 设置 y 轴标签
plt.ylabel('x1')
# 设置 x 轴标签
plt.xlabel('x0')
```

* 在上图中，蓝线应当表示直线 $x_0 + x_1 - 3 = 0$，它应当与 x1 轴相交于 3（若设 $x_1$ = 3，则 $x_0$ = 0），与 x0 轴相交于 3（若设 $x_1$ = 0，则 $x_0$ = 3）。


* 阴影区域表示 $-3 + x_0+x_1 < 0$。直线之上的区域是 $-3 + x_0+x_1 > 0$。


* 因此，对于这个模型，阴影区域内（直线下方）的任意点都被分类为 $y=0$，直线之上或直线上的任意点都被分类为 $y=1$。这条直线称为“决策边界”。

正如视频中所见，通过使用更高阶的多项式项（例如 $f(x) = g( x_0^2 + x_1 -1)$），我们可以得到更复杂的非线性边界。
