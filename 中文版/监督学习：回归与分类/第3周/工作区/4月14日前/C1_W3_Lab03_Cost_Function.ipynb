### 非向量化版本

回忆一下，对于逻辑回归，代价函数的形式为

$$ J(w) = \\frac{1}{m} \\sum_{i=1}^{m} \\left[ cost(f_{w}(x^{(i)}), y^{(i)}) \\right]$$

其中
* m 是数据集中的训练样本数

* $cost(f_{w}(x^{(i)}), y^{(i)})$ 是单个数据点的代价，即 -

    $$cost(f_{w}(x^{(i)}), y^{(i)}) = (-y^{(i)} \\log\\left(f_w\\left( x^{(i)} \\right) \\right) - \\left( 1 - y^{(i)}\\right) \\log \\left( 1 - f_w\\left( x^{(i)} \\right) \\right)$$
    
*  $f_{w}(x^{(i)})$ 是模型的预测，而 $y^{(i)}$ 是实际标签


* 对于上面给定数据集的逻辑回归模型，模型可以表示为：

    $f_{w}(x) = g(w_0 + w_1x_1 + w_2x_2)$

    其中 $g(z)$ 是 sigmoid 函数：

    $g(z) = \\frac{1}{1+e^{-z}}$ 


* **预处理步骤**

   为便于实现，我们给 $X$ 增加一列全为 1 的列（作为 $x_0$），从而  
    $f_{w}(x) = g(w_0x_0 + w_1x_1 + w_2x_2)$
    
    这样，要计算模型的预测 $f_{w}(x)$，我们可以写一个 for 循环，在每一步计算 $w_jx_j$
