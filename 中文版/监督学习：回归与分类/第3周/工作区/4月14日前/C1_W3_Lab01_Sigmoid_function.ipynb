# 不计分实验：Sigmoid 函数

在这个不计分实验中，你将实现 sigmoid 函数（也称为逻辑函数）。


正如讲课视频中所讨论的，对于分类任务，我们可以先使用原来的线性回归模型（$f(x) = w^Tx$），在给定 `X` 的情况下预测 `y`。
- 然而，由于我们的输出变量 `y` 取 0 或 1，我们希望分类模型的预测值在 0 和 1 之间。
- 这可以通过使用“sigmoid 函数”来实现，它把所有输入值映射到 0 和 1 之间的值。


让我们实现 sigmoid 函数，亲眼看看这一点。

## Sigmoid 函数的公式

sigmoid 函数的公式如下——

$g(z) = \\frac{1}{1+e^{-z}}$

在逻辑回归中，`z`（sigmoid 函数的输入）是线性回归模型的输出。
- 也就是说，`z` 并不总是一个数字，也可能是数字数组。
- 如果输入是数字数组，我们要对输入数组中的每个值应用 sigmoid 函数。

让我们用 Python 实现它。

 `numpy` 有一个名为 [`exp()`](https://numpy.org/doc/stable/reference/generated/numpy.exp.html) 的函数，它提供了一种便捷的方法来计算输入数组（`z`）中所有元素的指数（`e^{z}`）。

它也可以接受单个数字作为输入，如下所示。

```python
# 输入是数组。
input_array = np.array([1,2,3])
exp_array = np.exp(input_array)

print("Input to exp:", input_array)
print("Output of exp:", exp_array)

# 输入是单个数字
input_val = 1  
exp_val = np.exp(input_val)

print("Input to exp:", input_val)
print("Output of exp:", exp_val)
```

利用 `exp` 函数，我们可以把 sigmoid 函数公式 $g(z) = \\frac{1}{1+e^{-z}}$ 写成如下代码：

`g = 1/(1+np.exp(-z))`
