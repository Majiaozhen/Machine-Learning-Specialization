## 带正则化逻辑回归的代价函数

下面，我们将实现带正则化逻辑回归的代价函数。

回忆一下，对于带正则化的逻辑回归，代价函数的形式为
$$J(\\mathbf{w}) = \\frac{1}{m}  \\sum_{i=0}^{m-1} \\left[ -y^{(i)} \\log\\left(f_{\\mathbf{w}}\\left( \\mathbf{x}^{(i)} \\right) \\right) - \\left( 1 - y^{(i)}\\right) \\log \\left( 1 - f_{\\mathbf{w}}\\left( \\mathbf{x}^{(i)} \\right) \\right) \\right] + \\frac{\\lambda}{2m}  \\sum_{j=1}^{n} w_j^2$$

把它与不带正则化的代价函数（你在之前的实验中实现过）比较，后者形式为

$$ J(\\mathbf{w}) = \\frac{1}{m}\\sum_{i=0}^{m-1} \\left[ (-y^{(i)} \\log\\left(f_{\\mathbf{w}}\\left( \\mathbf{x}^{(i)} \\right) \\right) - \\left( 1 - y^{(i)}\\right) \\log \\left( 1 - f_{\\mathbf{w}}\\left( \\mathbf{x}^{(i)} \\right) \\right)\\right]$$

区别在于正则化项，即 $$\\frac{\\lambda}{2m}  \\sum_{j=1}^n w_j^2$$

注意，求和从 $j = 1$ 开始。也就是说，你不应对参数 $w_0$ 做正则化。

请完成 `compute_cost_reg` 函数，使其能够：
- 计算训练集中所有样本的代价
    - 遍历训练集中的所有样本。在循环外创建一个变量来存放总代价
    - 在循环内，完全按照你之前在 `compute_cost` 函数中的做法计算每个训练样本的代价——
        - 计算 `z`
        $$
        z =  w_0x_0^{(i)} + w_1x_1^{(i)} + w_2x_2^{(i)} = \\mathbf{w} \\cdot \\mathbf{x}^{(i)}
        $$
        - 预测 `f`，其中 `g` 是 sigmoid 函数
        $$
        f =  g(z)
        $$
        - 计算每个样本的代价

        $cost =  (-y^{(i)} * \\log(f)) - ((1 - y^{(i)})\\log( 1 - f))$
    
        - 把这个代价加到循环外创建的总代价变量上
        
        
- 通过遍历 `n`（特征数）计算正则化项。在循环外创建一个变量来存放正则化代价
    - 务必让循环从 1 开始（不是 0），因为你不希望用到 $w[0]$ 项
    - 在循环内计算 `w_j**2`
    - 把它加到正则化代价变量上
    
    
- 把两个循环的输出相加得到总和，即把
    - 所有训练样本的代价之和除以样本数
    - 遍历 $w$ 得到的正则化代价之和乘以 $\\frac{\\lambda}{2m}$

在做这件事时请记住，变量 X 和 y 不是标量值，而是形状分别为 ($m, n+1$) 和 ($𝑚$, ) 的矩阵，其中 $𝑛$ 是特征数，$𝑚$ 是训练样本数。

````python
def compute_cost_reg(X, y, w, lambda_=1): 
    """计算带正则化的逻辑回归代价。
   
    参数
    ----------
    X : array_like
        形状 (m, n+1)
   
    y : array_like
        形状 (m,)
   
    w : array_like
        模型参数
        形状 (n+1,)
       
    lambda_ : float
        控制正则化强度
        默认设为 1
   
    返回
    -------
    total_cost : float
        以 w 为带正则化的逻辑回归参数来拟合 X 和 y 中数据点的代价"""
    # 训练样本数
    m, n = X.shape
    
    ### 在此处编写代码 ###
    ### 解答开始 ###
    cost_list = []
    
    for i in range(m):
        z = np.dot(w.T, X[i])
        f = sigmoid(z)
        cost = -y[i]*np.log(f) - (1-y[i])*np.log(1-f)        
        cost_list.append(cost)
        
    total_cost = (1/m)* sum(cost_list)
    
    reg_cost = 0
    for j in range(1, n):
        reg_cost += (w[j]**2)
    
    total_cost += (lambda_/(2*m))*reg_cost
    ### 解答结束 ###
    ### 代码编写结束 ###
    
    return total_cost
````

运行下面的单元格，检查你对 `compute_cost_reg` 函数的实现。

````python
np.random.seed(1)
X_tmp = np.random.rand(5,6)
y_tmp = np.array([0,1,0,1,0])
initial_w = np.random.rand(X_tmp.shape[1])
lambda_ = 1
cost = compute_cost_reg(X_tmp, y_tmp, initial_w, lambda_)

print("Regularized cost at initial w:", cost)
````
