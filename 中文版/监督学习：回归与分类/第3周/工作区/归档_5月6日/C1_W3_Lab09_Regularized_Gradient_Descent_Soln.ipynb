### 带正则化逻辑回归的梯度

下面，你将实现带正则化逻辑回归的梯度。

带正则化代价函数的梯度是一个与参数 $\\mathbf{w}$ 形状相同的向量，其中第 $j^\\mathrm{th}$ 个元素定义如下：

$$\\frac{\\partial J(\\mathbf{w})}{\\partial w_0} = \\frac{1}{m}  \\sum_{i=0}^{m-1} (f_{\\mathbf{w}}(\\mathbf{x}^{(i)}) - y^{(i)}) x_j^{(i)} \\quad\\quad\\quad\\quad\\quad\\quad \\mbox{for $j = 0$}$$

$$\\frac{\\partial J(\\mathbf{w})}{\\partial w_j} = \\left( \\frac{1}{m}  \\sum_{i=0}^{m-1} (f_{\\mathbf{w}}(\\mathbf{x}^{(i)}) - y^{(i)}) x_j^{(i)} \\right) + \\frac{\\lambda}{m} w_j  \\quad\\, \\mbox{for $j \\geq 1$}$$


你将实现一个名为 `compute_gradient_reg` 的函数，返回 $\\frac{\\partial J(\\mathbf{w})}{\\partial \\mathbf{w}}$。

在下面的说明和代码中，术语 'dw' 指 $\\frac{\\partial J(\\mathbf{w})}{\\partial \\mathbf{w}}$。dJ 是隐含的。  
请完成 `compute_gradient_reg` 函数，使其能够：
- 遍历 dw（已初始化为 0）中的每个元素（`j`）。
    - 计算每个元素的梯度（$dw_j$）
        - 遍历训练集中的所有样本。在循环外创建一个变量来存放总梯度
        - 在循环内，计算每个训练样本带来的梯度更新
            - 计算 `z`
            $$
            z =  w_0x_0^{(i)} + w_1x_1^{(i)} + w_2x_2^{(i)} = \\mathbf{w} \\cdot \\mathbf{x}^{(i)}
            $$
            - 预测 `f`，其中 `g` 是 sigmoid 函数
            $$
            f =  g(z)
            $$
            - 计算每个样本带来的梯度

            $$gradient =  (f - y^{(i)})x_j^{(i)}$$
    
            - 把这个梯度加到循环外创建的总梯度变量上
    
        - 把 $dw_j$ 处的总梯度取为所有迭代梯度之和，并返回总和除以样本数。
    - 若 `j> 0`
        - 把 $\\frac{\\lambda}{m} w_j$ 加到 $dw_j$ 上

在做这件事时请记住，变量 X 和 y 不是标量值，而是形状分别为 ($m, n+1$) 和 ($𝑚$, ) 的矩阵，其中 $𝑛$ 是特征数，$𝑚$ 是训练样本数。


```python
## 计分函数：compute_gradient_reg

def compute_gradient_reg(X, y, w, lambda_=1): 
    """计算带正则化的逻辑回归梯度。
   
    参数
    ----------
    X : array_like
        形状 (m, n+1)
   
    y : array_like
        形状 (m,)
   
    w : array_like
        模型参数
        形状 (n+1,)
       
    lambda_ : float
        控制正则化强度
        默认设为 1
   
    返回
    -------
    dw : array_like
        形状 (n+1,)
        正则化代价对参数 w 的梯度。
        注意 dw 与 w 的维度相同。"""
    
    m, n = X.shape
    
    # 你需要替换/更新 dw
    dw = np.zeros_like(w)
    
    ### 在此处编写代码 ###
    ### 解答开始 ###
    for j in range(n):
        gradient_list = []
        
        for i in range(m):        
            z = np.dot(w.T, X[i])
            f = sigmoid(z)
            gradient = (f-y[i])*X[i][j]          
            gradient_list.append(gradient)
        
        dw[j] = (1/m)* sum(gradient_list)
        
        if j > 0:
            dw[j] += (lambda_/m)*w[j]
    ### 解答结束 ###
    ### 代码编写结束 ###
    
    return dw
```

运行下面的单元格，检查你对 `compute_gradient_reg` 函数的实现。

```python
np.random.seed(1)
X_tmp = np.random.rand(5,6)
y_tmp = np.array([0,1,0,1,0])
initial_w = np.random.rand(X_tmp.shape[1]) - 0.5
lambda_ = 1
grad = compute_gradient_reg(X_tmp, y_tmp, initial_w, lambda_)

print(f"First few elements of regularized gradient:\n {grad[:4]}", )
```
