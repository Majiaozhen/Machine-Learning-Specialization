## 1 - 实现 K-means

K-means 算法是一种把相似数据点自动聚成一类的方法。

* 具体来说，给定训练集 $\{x^{(1)}, ..., x^{(m)}\}$，你希望把数据分成若干紧凑的“簇（cluster）”。


* K-means 是一个迭代过程，它
 * 先猜测初始质心，然后
 * 通过以下方式不断改进这一猜测：
     * 反复把样本分配到离它最近的质心，然后
     * 根据这些分配结果重新计算质心。


* 用伪代码表示，K-means 算法如下：


``` python
# Initialize centroids
# K is the number of clusters
centroids = kMeans_init_centroids(X, K)

for iter in range(iterations):
    # Cluster assignment step: 
    # Assign each data point to the closest centroid. 
    # idx[i] corresponds to the index of the centroid 
    # assigned to example i
    idx = find_closest_centroids(X, centroids)

    # Move centroid step: 
    # Compute means based on centroid assignments
    centroids = compute_means(X, idx, K)
```



* 算法的内层循环反复执行两个步骤：
* (i) 把每个训练样本 $x^{(i)}$ 分配到离它最近的质心；
* (ii) 用分配给该质心的点重新计算该质心的均值。


* $K$-means 算法总会收敛到某一组最终的质心均值。

* 不过，收敛得到的解不一定总是理想的，它取决于质心的初始设置。
* 因此，实践中通常会以不同的随机初始化多次运行 K-means 算法。
* 在这些由不同随机初始化得到的不同解之间做选择的一种方法，是选取代价函数值（畸变，distortion）最低的那个。

你将在接下来的几节中分别实现 K-means 算法的两个阶段。
* 你会先完成 `find_closest_centroid`，然后再完成 `compute_centroids`。

## 2 - 在样本数据集上运行 K-means

完成上面两个函数（`find_closest_centroids` 和 `compute_centroids`）之后，下一步是在一个二维玩具数据集上运行 K-means 算法，以帮助你理解 K-means 的工作方式。

````python
# 这一部分你不需要实现任何内容

def run_kMeans(X, initial_centroids, max_iters=10, plot_progress=False):
    """在数据矩阵 X 上运行 K-Means 算法，其中 X 的每一行
   是一个样本"""
    
    # 初始化取值
    m, n = X.shape
    K = initial_centroids.shape[0]
    centroids = initial_centroids
    previous_centroids = centroids    
    idx = np.zeros(m)
    
    # 运行 K-Means
    for i in range(max_iters):
        
        #输出进度
        print("K-Means iteration %d/%d" % (i, max_iters-1))
        
        # 对 X 中的每个样本，把它分配给最近的质心
        idx = find_closest_centroids(X, centroids)
        
        # 可选地绘制进度
        if plot_progress:
            plot_progress_kMeans(X, centroids, previous_centroids, idx, K, i)
            previous_centroids = centroids
            
        # 根据归属关系计算新的质心
        centroids = compute_centroids(X, idx, K)
    plt.show() 
    return centroids, idx
````

````python
# 加载一个示例数据集
X = load_data()

# 设置初始质心
initial_centroids = np.array([[3,3],[6,2],[8,5]])
K = 3

# 迭代次数
max_iters = 10

centroids, idx = run_kMeans(X, initial_centroids, max_iters, plot_progress=True)
````

## 3 - 随机初始化

示例数据集的质心初始分配是特意设计的，以便你看到与图 1 相同的图形。实践中，初始化质心的一个好策略是从训练集中随机选取样本。

在练习的这一部分，你应理解函数 `kMeans_init_centroids` 是如何实现的。
* 代码首先随机打乱样本的索引（使用 `np.random.permutation()`）。
* 然后，根据索引的随机排列选出前 $K$ 个样本。
    * 这样就能随机选取样本，且不会出现同一个样本被选两次的风险。

**注意**：练习的这一部分不需要你实现任何内容。

````python
# 这一部分你不需要修改

def kMeans_init_centroids(X, K):
    """本函数初始化用于在数据集 X 上做 K-Means 的 K 个质心
   
    参数：
        X (ndarray): 数据点
        K (int):     质心/簇的数量
   
    返回：
        centroids (ndarray): 初始化后的质心"""
    
    # 随机打乱样本的索引顺序
    randidx = np.random.permutation(X.shape[0])
    
    # 取前 K 个样本作为质心
    centroids = X[randidx[:K]]
    
    return centroids
````
